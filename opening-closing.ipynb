```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def load_img():
    blank_img = np.zeros((600, 850), dtype=np.uint8)
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(blank_img,text='LOHINI',org=(175, 359),fontFace=font,fontScale=5,color=255,thickness=25,lineType=cv2.LINE_AA)
    return blank_img
```

```python
img = load_img()
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
noisy_img = img.copy()

noise = np.random.randint(0, 100, (600, 850))

noisy_img[noise > 90] = 255

plt.figure(figsize=(10, 7))
plt.imshow(noisy_img, cmap='gray')
plt.title("Original Image with Noise")
plt.axis("off")
plt.show()
```

```python
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
opened_img = cv2.morphologyEx(noisy_img,cv2.MORPH_OPEN,kernel)
print(kernel)

plt.figure(figsize=(10, 7))
plt.imshow(opened_img, cmap='gray')
plt.title("After Opening")
plt.axis("off")
plt.show()
```

```python
hole_img = img.copy()

cv2.circle(hole_img, (150, 260), 8, 0, -1)
cv2.circle(hole_img, (300, 280), 10, 0, -1)
cv2.circle(hole_img, (500, 250), 8, 0, -1)
cv2.circle(hole_img, (700, 280), 10, 0, -1)

plt.figure(figsize=(10, 7))
plt.imshow(hole_img, cmap='gray')
plt.title("Image with Holes")
plt.axis("off")
plt.show()
```

```python
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (15, 15))

closed_img = cv2.morphologyEx(hole_img,cv2.MORPH_CLOSE,kernel)

plt.figure(figsize=(10, 7))
plt.imshow(closed_img, cmap='gray')
plt.title("After Closing")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(noisy_img, cmap='gray')
plt.title("Noisy Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(opened_img, cmap='gray')
plt.title("Opening")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(closed_img, cmap='gray')
plt.title("Closing")
plt.axis("off")

plt.tight_layout()
plt.show()
```

# MORPHOLOGICAL GRADIENT

```python
gradient_img = cv2.morphologyEx(img,cv2.MORPH_GRADIENT,kernel)

plt.figure(figsize=(10, 7))
plt.imshow(gradient_img, cmap='gray')
plt.title("Morphological Gradient")
plt.axis("off")
plt.show()
```
